# 2. Baza Teorike

## 2.1 Oshiloskopi

Oshiloskopi është një instrument elektronik që përdoret për të paraqitur dhe matur sinjalet elektrike në funksion të kohës.

Ai na mundëson të studiojmë formën e sinjalit, amplitudën, periudhën dhe frekuencën.

Parametrat kryesorë të oshiloskopit janë:

- *TIME/DIV:* Përcakton kohën për çdo ndarje horizontale.
- *VOLT/DIV:* Përcakton tensionin për çdo ndarje vertikale.
- *CH1 dhe CH2:* Kanalet e matjes së sinjaleve.
- *TRIGGER:* Përdoret për stabilizimin e sinjalit në ekran.

### Formulat kryesore

Periudha:

$$
T=N_h \\cdot TIME/DIV
$$

Tensioni:

$$
V=N_v \\cdot VOLT/DIV
$$

Frekuenca:

$$
f=\\frac{1}{T}
$$

Ku:

- $T$ – periudha në sekonda.
- $f$ – frekuenca në Hertz.
- $N_h$ – numri i ndarjeve horizontale.
- $N_v$ – numri i ndarjeve vertikale.

# 2.2 Qarqet RC

Qarku RC është një qark elektrik i përbërë nga një rezistencë (R) dhe një kondensator (C).

Ky qark përdoret për të studiuar proceset e karikimit dhe shkarkimit të kondensatorit.

### Karikimi i kondensatorit

Kur kondensatori lidhet me një burim tensioni përmes një rezistence, ai fillon të karikohet.

Tensioni gjatë karikimit jepet nga formula:

$$
V_C(t)=V(1-e^{-t/RC})
$$

### Shkarkimi i kondensatorit

Gjatë shkarkimit, tensioni në kondensator zvogëlohet në mënyrë eksponenciale:

$$
V_C(t)=V_0e^{-t/RC}
$$

### Konstanta kohore

Konstanta kohore e qarkut RC përcaktohet nga:

$$
\\tau=R\\cdot C
$$

Ku:

- $R$ – rezistenca në Ohm (Ω).
- $C$ – kapaciteti në Farad (F).
- $\\tau$ – konstanta kohore në sekonda (s).

Pas një konstanteje kohore, kondensatori arrin rreth 63.2% të tensionit përfundimtar gjatë karikimit.

$$
T=N_h\\cdot TIME/DIV

$$
T=5.4\\cdot0.1
$$

$$
\\boxed{T=0.54\\ s}
$$

## 5.2 Matjet vertikale (VOLT/DIV)

$$
V=N_v\\cdot VOLT/DIV
$$

$$
V=8\\cdot0.5
$$

$$
\\boxed{V=4\\ V}
$$

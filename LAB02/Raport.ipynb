# 2. Baza Teorike

Rezistenca elektrike është madhësia fizike që tregon kundërshtimin ndaj kalimit të rrymës elektrike.

Njësia matëse e saj është Ohmi (Ω).

Ligji i Ohmit:

$$
U=R\cdot I
$$

Ku:

- U – Tensioni elektrik (V)
- R – Rezistenca (Ω)
- I – Intensiteti i rrymës (A)

### Ndarësi i tensionit

Kur dy rezistenca lidhen në seri, tensioni ndahet ndërmjet tyre.

Formula e tensionit në dalje është:

$$
V_{out}=V_{in}\cdot\frac{R_2}{R_1+R_2}
$$

Breadboard përdoret për ndërtimin dhe testimin e qarqeve elektrike pa pasur nevojë për saldim.

# 4. Eksperimenti i Parë

## Matja me rezistenca në rendin kΩ

### Të dhënat eksperimentale

| Madhësia | Vlera |
|---|---:|
| R1 | 1,781 kΩ |
| R2 | 1,765 kΩ |
| Vin | 11,36 V |
| Vout | 5,68 V |

### Llogaritjet

Raporti i tensionit në dalje me tensionin në hyrje:

$$
\frac{V_{out}}{V_{in}}=
\frac{5,68}{11,36}=0,5
$$

Raporti i rezistencave:

$$
\frac{R_2}{R_1+R_2}
=
\frac{1,765}{1,781+1,765}
\approx 0,502
$$

### Analiza

Nga rezultatet vërejmë se rezistencat R1 dhe R2 kanë vlera pothuajse të barabarta. Për këtë arsye, tensioni në dalje është afërsisht gjysma e tensionit në hyrje.

# 5. Eksperimenti i Dytë

## Matja me rezistenca në rendin MΩ

### Të dhënat eksperimentale

| Madhësia | Vlera |
|---|---:|
| R1 | 3,264 MΩ |
| R2 | 3,229 MΩ |
| Vin | 11,36 V |
| Vout | 4,83 V |

### Llogaritjet

Raporti eksperimental i tensionit:

$$
\frac{V_{out}}{V_{in}}
=
\frac{4,83}{11,36}
\approx 0,425
$$

Raporti teorik i rezistencave:

$$
\frac{R_2}{R_1+R_2}
=
\frac{3,229}{3,264+3,229}
\approx 0,495
$$

### Analiza

Në këtë eksperiment përdorëm rezistenca me vlera në megaohm (MΩ).

Nga matjet vërejmë një ndryshim ndërmjet raportit teorik dhe atij eksperimental të tensionit.

Ky ndryshim mund të lidhet me tolerancën e komponentëve, ngarkimin e qarkut dhe saktësinë e matjeve.

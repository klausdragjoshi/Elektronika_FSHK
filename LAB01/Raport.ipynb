## Ligji i Ohmit

Ligji i Ohmit shpreh lidhjen ndërmjet tensionit, rrymës dhe rezistencës elektrike.

Formula matematikore:

$$V = I \cdot R$$

Ku:

- V = Tensioni (Volt)
- I = Intensiteti i rrymës (Amper)
- R = Rezistenca (Ohm)

## Shembull praktik

Të dhënat:

- V = 12 V
- R = 4 Ω

Formula:

$$I = \frac{V}{R}$$

Zëvendësojmë vlerat:

$$I = \frac{12}{4} = 3 A$$

Përgjigjja: Intensiteti i rrymës është 3 Amper
